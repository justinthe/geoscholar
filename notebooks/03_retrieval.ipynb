```python
# Cell 1: build every component once
from geoscholar.index import Embedder, Index
from geoscholar.retrieval import Retriever, Reranker, tokenize

embedder = Embedder()
retriever = Retriever(Index(embedder), reranker=Reranker())
print(len(retriever.bm25.payloads), "chunks in BM25")
print(tokenize("Flood mapping with Sentinel-1 SAR and U-Net"))
```

```python
# Cell 2: same question, five strategies
def compare(q, k=3):
    print("Q:", q)
    for mode, rr in [("dense", False), ("bm25", False), ("hybrid", False), ("dense", True), ("hybrid", True)]:
        print(f"\n  {mode}{' + rerank' if rr else ''}")
        for score, h in retriever.search(q, k=k, mode=mode, rerank=rr):
            print(f"    {score:7.3f}  {h['paper_id']}  [{h['section'][:25]}]  {h['title'][:50]}")

compare("LEVIR-CD")                                                   # exact term
compare("using radar to see the ground through clouds")              # paraphrase
```
